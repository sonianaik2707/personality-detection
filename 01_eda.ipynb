```python
import pandas as pd

df = pd.read_csv('mbti_1.csv')

print('Shape:', df.shape)
print(df.head())
print(df['type'].value_counts())
```

```python
df['label'] = df['type'].apply(lambda x: 'Introvert' if x[0] == 'I' else 'Extrovert')
print(df['label'].value_counts())
print(df['label'].value_counts(normalize=True).round(2)) # class balance %
```

```python
plt.figure(figsize=(6,4))
df['label'].value_counts().plot(kind='bar', color=['#185FA5','#0F6E56'],
edgecolor='white')
plt.title('Class Distribution')
plt.xlabel('Personality Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('class_dist.png', dpi=150) # save for README later
plt.show()
```
